Roboflow doesnt support downloading from universe, so fork it then give the id.

```python
import requests
import os
from dotenv import load_dotenv
from roboflow import Roboflow

load_dotenv()

API_KEY = os.getenv("API_KEY")

rf = Roboflow(API_KEY)

# change to the project id you wanna download 
project = rf.project("pores1-5-2-ohjfg")
```

```python
records = []

for page in project.search_all(
    offset=0,
    limit=100,
    in_dataset=True,
    batch=False,
    fields=["id", "name", "owner"],
):
    records.extend(page)

project_name = project.name

output_dir = os.path.join("raw_images", project_name)
os.makedirs(output_dir, exist_ok=True)

for record in records:
    url = f"https://source.roboflow.com/{record['owner']}/{record['id']}/original.jpg"

    response = requests.get(url)
    response.raise_for_status()

    path = os.path.join(output_dir, record["name"])

    with open(path, "wb") as f:
        f.write(response.content)

    print(f"Downloaded: {record['name']}")
```
